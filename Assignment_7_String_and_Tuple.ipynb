#2. String Slicing and Indexing:

## a. Print the first character of the string.

 We can use index method to print the first character

```python
print(result1[0])
```

The concatenated string from the above output is "Hello Zara , welcome to python programming".So the first characher is "H".Here the first character is in the index position 0, so we can print it using index method.

## b. Print the last character of the string.

```python
print(result1[-1])
```

We can use negative index method to print the last character of the given input, Here there is no need to count the index of each character.Index position [-1] will give the output as the last character of the input.

## c. Print the first 5 characters of the string.

```python
print(result1[0:5])
```

## d. Print the last 11 characters of the string.

```python
print(result1[-11:])
```

Here we have to print the last 11 characters, we can use slicing method,No need to mention the last character since if we mention the last character inside the slicing automatically it will print one character before.

## e. Print the string in reverse.

```python
print(result1[::-1])
```

Here we can do the reversing using slice method, this is the easiest method to reverse the input.
No need to add the start and stop index, only add stepsize , it will reverse the input, -1 is the last character of the input.
